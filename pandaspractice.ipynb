```python
#Create a series from a list
import pandas as pd
a = [1,7,2]
myvar = pd.Series(a)
print(myvar)
```

```python
#Create labels
a = [1,7,2]
myvar = pd.Series(a, index = ["x","y","z"])
print(myvar)
```

```python
#naming a series
s1 = pd.Series([10,20,30], index=['a','b','c'], name = 'Example Series')
print(s1)
```

```python
#create a series from a dictionary
s2 = pd.Series({'x':1,'y':2,'z':3})
print(s2)
```

```python
#Named indexes
import pandas as pd 
data= {"calories" : [420,380,390],
    "duration": [50,40,45]
}
df = pd.DataFrame(data, index = ["day1","day2","day3"])
print(df)
```

```python
import pandas as pd
data = {
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,None,5]
}
df = pd.DataFrame(data)
print("Orignal Data:\n",df)
print()

#use dropna() to remove rows with any missing values
df_cleaned = df.dropna()
print(df_cleaned)
```

```python
#Fill missing values
import pandas as pd
data = {
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,None,5]
}
df = pd.DataFrame(data)
print("Orignal Data:\n",df)

#filling NaN values with 0
df.fillna(0,inplace=True)
```

```python
import pandas as pd
#sample data
data = {
    'A' : [1,2,2,3,3,4]
    'B' : [5,6,6,7,8,8]
}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df.to_string(index=False))
#detect duplicates
print("Duplicate Rows:\n", df[df.duplicated()].to_string(index = False))
```
