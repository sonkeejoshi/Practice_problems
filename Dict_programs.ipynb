```python
my_dict = {0:10, 1:20} 
my_dict[2] = 30
print("Dictionary:", my_dict)
```

```python
dic1 = {1:10, 2:30}
dic2 = {3:30, 4:40}
dic3 = {5:50, 6:60}
new_dict = {}
new_dict.update(dic1)
new_dict.update(dic2)
new_dict.update(dic3)
print("New dictionary", new_dict)
```

```python
my_dict = {101:"Amit", 102: "Rahul", 103: "Priya"}
key = int(input("Enter key to search:"))
if key in my_dict:
    print("Key exists")
else:
    print("Key doesnt exist")
```

```python
#keys and values
my_dict= {
    101:"Amit", 
    102: "Rahul", 
    103: "Priya"
}
print("Keys:")
for key in my_dict:
    print(key)
print("\nValues:")
for value in my_dict.values():
    print(value)
print("\nKeys and Values:")
for key, value in my_dict.items():
    print(key,value)
```

```python
my_dict = {}
for i in range(1,16):
    my_dict[i] = i**2
print(my_dict)
```

```python
#sum of values
my_dict ={"a":10, "b":20, "c":30, "d":40}
total = 0
for value in my_dict.values():
 total = total + value
print("Sum of values:", total)
```

```python
#add and delete 
my_dict = {
    101:"Amit", 
    102: "Rahul", 
    103: "Priya",
    104: "Popo",
    105: "Lolo"
}
print("Dictionary: ", my_dict)
del my_dict[101]
my_dict[106] = "Zozo"

print("Updated dict:", my_dict)
```

```python
#convert list to dict
list1 = ["Name", "Panel", "Roll no"]
list2 = ["ABC", "B", 34]
my_dict = dict(zip(list1, list2))
print("Dictionary:", my_dict)
```
