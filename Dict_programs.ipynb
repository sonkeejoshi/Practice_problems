```python
#sorting
my_dict={
    "name":"Popo",
    "panel":"B",
    "marks":[89,78,97,87]
}
sorted_dict=sorted(my_dict)
print("Sorted keys:", sorted_dict)
```

```python
#display student with highest marks
students={
    "Amit":78,
    "Rahul":92,
    "Sneha":88
}
highest_student = max(students, key=students.get)
print("Student with highest marks is ",highest_student)
print("Highest Marks: ", students[highest_student])
```

```python
#count frequency
numbers = [1,2,1,3,4,2,1,2]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num]+1
    else:
        frequency[num] = 1
    print("Frequency of elements: ",frequency)
```

```python
#sort dictionary by values
my_dict = {
     "Amit":78,
    "Rahul":92,
    "Sneha":88
}
sorted_dict =  dict(sorted(my_dict.items(),key=lambda item :item[1]))
print("Dictionary sorted by values ",sorted_dict)
```

```python
#sort dictionary by values(descending)
my_dict = {
     "Amit":78,
    "Rahul":92,
    "Sneha":88
}
sorted_dict =  dict(sorted(my_dict.items(),key=lambda item :item[1], reverse=True))
print("Dictionary sorted by values ",sorted_dict)
```
